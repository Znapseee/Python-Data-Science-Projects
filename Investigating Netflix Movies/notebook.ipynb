```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

```python
# Start the project by having an overview of the dataset
netflix_df.head()
```

# Filter the data for movies released in the 1990s
You'll want to subset the DataFrame to keep only the movies and then filter the years so that you are working with rows where the release year is between 1990 and 1999.

```python
# Filter the type from the netflix_df dataset
movies = netflix_df[netflix_df["type"] == 'Movie']

# Filter the movies between the year 1990 to 1999
movie_90s = movies[(movies['release_year'] >= 1990) & (movies['release_year'] <= 1999)]

print(movie_90s.head())
```

# Find the Most Frequent Movie Duration
Visualize the data to view the distributions of durations for movies released in the 1990s.

```python
# Extract numeric duration from the 'duration' column if it's not already an integer
duration = movie_90s['duration']

# Plotting the data using a histogram
plt.title('Distribution of the Movies Released in 1990s')
plt.hist(duration, bins=10)
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')
plt.show()

duration = 100
```

# Count the number of Short Action Movies from the 1990s

First, subset the data again to keep only movies with the correct genre. Then, iterate through the data and count the number of movies depending on the condition pf a duration that is less than 90.

```python
# Filter the Action genre from the dataset
action_movies = netflix_df['genre'] == 'Action'

# Filter the Action movies from the 1990s
action_movie_90s = movie_90s[action_movies]
action_movie_90s.head()

# Setting up a counter
short_movie_count = 0

for label, row in action_movie_90s.iterrows():
    if row['duration'] < 90:
        short_movie_count += 1
    else:
        short_movie_count = short_movie_count

print(short_movie_count)
```
